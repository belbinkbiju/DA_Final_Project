## Project Objective

## Business Context Clarification

During the data quality investigation, the meaning of certain fields was clarified with internal product knowledge.

The `account_status` field should not be used alone to identify invalid or incomplete customer accounts. `UNVERIFIED` accounts with one or more children can represent genuine parent accounts whose email has been verified but whose phone number verification is pending. Phone verification is completed when the parent installs the application and logs in to activate the card(s). Therefore, `account_status` will be treated as a descriptive customer-account attribute rather than an exclusion criterion.

Similarly, `number_of_children` represents the number of child accounts associated with a parent and does not necessarily represent the number of currently active cards. This particular prepaid card permits a maximum of four active cards at one time, but previously cancelled child cards may allow a parent to add additional children later. Therefore, parents with more than four historical child accounts are valid accounts and should not be excluded.

In this project, the term "parent" refers to the adult account holder who applies for and manages the account. This person may be a biological parent, grandparent, elder sibling, or another adult aged 18 or above.

## Defining the Analysis Population

For this analysis, the focus is on parent accounts associated with at least one child account. Records with `number_of_children = 0` will be excluded from the analysis population because they are outside the scope of this customer-card engagement analysis. Based on internal business validation, these records primarily represent incomplete applications or gifting accounts.

**The original dataset was preserved as `df`. A separate `analysis_df` will be created for subsequent analysis.**

The `card_activation_date` column will also be excluded from `analysis_df` because data validation identified inconsistencies between activation dates and first top-up dates. The first top-up date will be retained as the earliest reliable post-creation engagement milestone.

```python
# Detect outliers using the IQR method
outlier_summary = []

for column in numerical_columns:
    Q1 = analysis_df[column].quantile(0.25)
    Q3 = analysis_df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (analysis_df[column] < lower_bound) |
        (analysis_df[column] > upper_bound)
    ).sum()

    outlier_summary.append({
        'Column': column,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': outlier_count
    })

outlier_summary_df = pd.DataFrame(outlier_summary)

outlier_summary_df
```

### Feature Engineering: Customer Segmentation

Three categorical features will be created to analyse customer progression and transaction behaviour.

| Feature | Category | Definition |
|---|---|---|
| Customer Lifecycle Stage | `Created_No_Topup` | Account created, but no first top-up recorded. |
|  | `Topup_No_Transaction` | First top-up recorded, but no first transaction recorded. |
|  | `First_Transaction` | One recorded transaction. |
|  | `Repeat_Transaction` | More than one recorded transaction. |
| Transaction Activity Segment | `No_Activity` | 0 recorded transactions. |
|  | `Low_Activity` | 1–5 recorded transactions. |
|  | `Medium_Activity` | 6–20 recorded transactions. |
|  | `High_Activity` | More than 20 recorded transactions. |
| Transaction Value Segment | `No_Value` | £0 recorded transaction value. |
|  | `Low_Value` | Greater than £0 and up to £50. |
|  | `Medium_Value` | Greater than £50 and up to £200. |
|  | `High_Value` | Greater than £200. |

> The segmentation thresholds are initial analytical categories and may be reviewed during EDA. All categories are based on recorded historical activity and do not indicate current account status.

```python
def assign_lifecycle_stage(row):
    if not row['has_first_topup']:
        return 'Created_No_Topup'
    elif not row['has_first_transaction']:
        return 'Topup_No_Transaction'
    elif row['total_transaction_count'] == 1:
        return 'First_Transaction'
    else:
        return 'Repeat_Transaction'

analysis_df['customer_lifecycle_stage'] = analysis_df.apply(
    assign_lifecycle_stage,
    axis=1
)
```

```python
def assign_activity_segment(count):
    if count == 0:
        return 'No_Activity'
    elif count <= 5:
        return 'Low_Activity'
    elif count <= 20:
        return 'Medium_Activity'
    else:
        return 'High_Activity'

analysis_df['transaction_activity_segment'] = (
    analysis_df['total_transaction_count']
    .apply(assign_activity_segment)
)
```

```python
def assign_value_segment(value):
    if value == 0:
        return 'No_Value'
    elif value <= 50:
        return 'Low_Value'
    elif value <= 200:
        return 'Medium_Value'
    else:
        return 'High_Value'

analysis_df['transaction_value_segment'] = (
    analysis_df['total_transaction_value']
    .apply(assign_value_segment)
)
```

#### 🔎 Step 36: Top-up Behaviour by Product Type & Account Status

```python
# Step 36.1: Top-up Behaviour by Product Type

topup_by_product = pd.crosstab( #pd.crosstab() is a Pandas function used to create a frequency table between categorical variables.
    df['product_type'],
    df['has_first_topup']
)

topup_by_product
```

```python
topup_by_product_percentage = (
    topup_by_product
    .div(topup_by_product.sum(axis=1), axis=0)
    .mul(100)
    .round(2)
)

topup_by_product_percentage
```

```python
# Step 36.2: Top-up Behaviour by Account Status

topup_by_status = pd.crosstab(
    df['account_status'],
    df['has_first_topup']
)

topup_by_status
```

```python
topup_by_status_percentage = (
    topup_by_status
    .div(topup_by_status.sum(axis=1), axis=0)
    .mul(100)
    .round(2)
)

topup_by_status_percentage
```

```python
# Step 36.3: Top-up Behaviour by Product Type

plt.figure(figsize=(10, 5))

product_order = ['FDD', 'SDD']

plot_data = topup_by_product.loc[product_order]

fig, ax = plt.subplots(figsize=(10, 5))

y = np.arange(len(product_order))
bar_height = 0.35

no_topup = plot_data[False].values
topup_completed = plot_data[True].values

no_topup_pct = topup_by_product_percentage.loc[product_order, False].values
topup_completed_pct = topup_by_product_percentage.loc[product_order, True].values

ax.barh(
    y + bar_height / 2,
    no_topup,
    height=bar_height,
    color='#E76F51',
    label='No Top-up'
)

ax.barh(
    y - bar_height / 2,
    topup_completed,
    height=bar_height,
    color='#2A9D8F',
    label='Top-up Completed'
)

for i, (value, percentage) in enumerate(zip(no_topup, no_topup_pct)):
    ax.text(
        value + 30,
        y[i] + bar_height / 2,
        f'{value:,} ({percentage:.2f}%)',
        va='center'
    )

for i, (value, percentage) in enumerate(zip(topup_completed, topup_completed_pct)):
    ax.text(
        value + 30,
        y[i] - bar_height / 2,
        f'{value:,} ({percentage:.2f}%)',
        va='center'
    )

ax.set_yticks(y)
ax.set_yticklabels(product_order)
ax.set_xlabel('Number of Accounts')
ax.set_ylabel('Product Type')
ax.set_title('Top-up Behaviour by Product Type')

ax.legend()

ax.invert_yaxis()

plt.grid(False)
plt.tight_layout()
plt.show()
```

```python
# Step 36.4: Top-up Behaviour by Account Status

product_order = ['UNVERIFIED', 'VERIFIED']

plot_data = topup_by_status.loc[product_order]

fig, ax = plt.subplots(figsize=(10, 5))

y = np.arange(len(product_order))
bar_height = 0.35

no_topup = plot_data[False].values
topup_completed = plot_data[True].values

no_topup_pct = topup_by_status_percentage.loc[product_order, False].values
topup_completed_pct = topup_by_status_percentage.loc[product_order, True].values

ax.barh(
    y + bar_height / 2,
    no_topup,
    height=bar_height,
    color='#E76F51',
    label='No Top-up'
)

ax.barh(
    y - bar_height / 2,
    topup_completed,
    height=bar_height,
    color='#2A9D8F',
    label='Top-up Completed'
)

for i, (value, percentage) in enumerate(zip(no_topup, no_topup_pct)):
    ax.text(
        value + 30,
        y[i] + bar_height / 2,
        f'{value:,} ({percentage:.2f}%)',
        va='center'
    )

for i, (value, percentage) in enumerate(zip(topup_completed, topup_completed_pct)):
    ax.text(
        value + 30,
        y[i] - bar_height / 2,
        f'{value:,} ({percentage:.2f}%)',
        va='center'
    )

ax.set_yticks(y)
ax.set_yticklabels(product_order)
ax.set_xlabel('Number of Accounts')
ax.set_ylabel('Account Status')
ax.set_title('Top-up Behaviour by Account Status')

ax.legend()

ax.invert_yaxis()

plt.grid(False)
plt.tight_layout()
plt.show()
```

#### 🔎 Step 40: Time / Cohort Analysis

```python
# Step 40.1: Customer Behaviour by Creation Cohort

cohort_summary = (
    df.groupby('creation_year')
    .agg(
        total_accounts=('account_id', 'count'),
        topup_accounts=('has_first_topup', 'sum'),
        transaction_active_accounts=('total_transaction_count', lambda x: (x > 0).sum()),
        median_transaction_count=('total_transaction_count', 'median'),
        median_transaction_value=('total_transaction_value', 'median')
    )
)

# Calculate percentages
cohort_summary['topup_rate'] = (
    cohort_summary['topup_accounts']
    / cohort_summary['total_accounts'] * 100
).round(2)

cohort_summary['transaction_active_rate'] = (
    cohort_summary['transaction_active_accounts']
    / cohort_summary['total_accounts'] * 100
).round(2)

cohort_summary
```

```python
# Step 40.2: Cohort engagement percentages

cohort_engagement = cohort_summary[
    ['total_accounts', 'topup_rate', 'transaction_active_rate']
].copy()

cohort_engagement
```

```python
# Step 40.3: Cohort Engagement Visualization

plt.figure(figsize=(10, 6))

plt.plot(
    cohort_engagement.index,
    cohort_engagement['topup_rate'],
    marker='o',
    label='Top-up Rate'
)

plt.plot(
    cohort_engagement.index,
    cohort_engagement['transaction_active_rate'],
    marker='o',
    label='Transaction Active Rate'
)

plt.title('Customer Engagement by Account Creation Cohort',
          fontsize=16, fontweight='bold')

plt.xlabel('Account Creation Year')
plt.ylabel('Percentage of Accounts')

plt.xticks(cohort_engagement.index)
plt.ylim(0, 100)

plt.grid(axis='y', alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```

##### 📝Observations

- Customer engagement varies across account-creation cohorts, with both **top-up rate** and **transaction-active rate** generally declining from the earlier cohorts toward the more recent cohorts.

- The **2020 cohort shows the highest engagement**, with a **91.20% top-up rate** and **88.91% transaction-active rate**.

- Engagement gradually declined through 2021–2023. The **2023 cohort recorded 78.60% top-up rate and 75.79% transaction-active rate**, representing the lowest levels before the 2024 rebound.

- The **2024 cohort shows a noticeable improvement**, with the top-up rate increasing to **82.48%** and the transaction-active rate to **80.34%**.

- The **2025 cohort shows another decline**, with **76.43%** of accounts topping up and **72.61%** becoming transaction-active.

- The **2026 cohort has a slightly higher top-up rate (77.23%) than 2025**, but its transaction-active rate is slightly lower at **72.28%**.

- Across all cohorts, the **transaction-active rate remains below the top-up rate**, indicating that some customers who complete a top-up do not necessarily proceed to transaction activity.

- The recent cohorts, particularly **2025 and 2026, should be interpreted cautiously**, because newer accounts have had less time to progress through the customer lifecycle.

- Overall, the cohort analysis suggests that **customer engagement has weakened among more recent account cohorts compared with the earlier years**, although the pattern is not strictly linear and includes a clear improvement in 2024.

# 📌 Phase 6: Final Insights & Recommendations

The exploratory data analysis examined account creation, customer lifecycle progression, top-up behaviour, transaction activity, transaction value, product engagement, and account-creation cohorts.

The following insights represent the strongest and most relevant findings from the analysis.

---

## 🔎 Key Insight 1: Most Accounts Progress to Engagement, but a Meaningful Inactive Segment Remains

The majority of accounts progressed beyond account creation and became financially active.

- **87.08%** of accounts completed at least one recorded top-up.
- **84.74%** of accounts were transaction-active.
- **12.92%** of accounts had no recorded top-up.
- **15.26%** of accounts had no recorded transactions.
- `Repeat_Transaction` was the largest lifecycle stage, containing **82.42%** of accounts.

This indicates that most accounts eventually engage with the service, but a meaningful minority stop before reaching important stages of the customer journey.

The largest identifiable drop-off occurs at the early engagement stage, where accounts are created but do not complete a recorded first top-up.

**Key takeaway:**  
> The primary engagement opportunity appears to be converting newly created accounts into active customers, particularly before or around the first top-up stage.

---

## 🔎 Key Insight 2: Earlier First Top-up is Associated with Higher Transaction Engagement

The analysis of top-up timing shows a clear difference between transaction-active accounts and accounts with no transactions.

- Transaction-active accounts had a **mean time to first top-up of 14.90 days**.
- Accounts with no transactions had a **mean of 24.54 days**.
- The median was **9 days** for transaction-active accounts compared with **11.5 days** for accounts with no transactions.

Both groups contained delayed top-up outliers, but the overall pattern indicates that customers who top up earlier are more likely to become transaction-active.

This analysis identifies an association rather than proving that earlier top-up directly causes higher transaction activity.

**Key takeaway:**  
> Faster conversion from account creation to first top-up is associated with stronger downstream transaction engagement.

---

## 🔎 Key Insight 3: Transaction Frequency is Strongly Associated with Transaction Value

The relationship analysis found a strong positive correlation between total transaction count and total transaction value.

- Correlation coefficient: **0.856**
- Accounts with more transactions generally recorded higher total transaction values.
- The log-scale visualization confirmed the relationship while making the pattern easier to observe despite extreme values.

There are still accounts with unusually high transaction counts or values, indicating the presence of outliers.

**Key takeaway:**  
> Transaction frequency is strongly associated with the total value generated by an account, making sustained transaction activity an important indicator of customer value.

---

## 🔎 Key Insight 4: Product Type Shows Very Little Difference in Transaction Engagement

The analysis found almost identical transaction-active rates for FDD and SDD accounts.

- **FDD:** 84.71% transaction-active
- **SDD:** 84.86% transaction-active
- Difference: only **0.15 percentage points**

Top-up behaviour was also very similar:

- **FDD:** 87.11% completed a top-up
- **SDD:** 86.96% completed a top-up

Therefore, product type does not appear to be a major differentiator of customer engagement within this dataset.

**Key takeaway:**  
> Customer engagement is remarkably similar across FDD and SDD, suggesting that other factors may be more important in explaining differences in customer behaviour.

---

## 🔎 Key Insight 5: Account Verification Status Shows a Large Difference in Top-up Behaviour

A substantial difference was observed when comparing verified and unverified accounts.

- **89.13%** of verified accounts completed a top-up.
- Only **2.21%** of unverified accounts completed a top-up.
- **97.79%** of unverified accounts had no recorded top-up.

This suggests a strong association between account verification and progression to the top-up stage.

However, the unverified group contains only **181 accounts**, compared with **7,491 verified accounts**. Therefore, the difference should be interpreted with caution because of the substantial imbalance between the groups.

**Key takeaway:**  
> Verification status is strongly associated with top-up completion, although the relatively small unverified population limits how broadly this finding should be interpreted.

---

## 🔎 Key Insight 6: Customer Engagement is Generally Lower in More Recent Cohorts

The cohort analysis shows that both top-up rate and transaction-active rate were generally higher among earlier account-creation cohorts.

The **2020 cohort** recorded the highest engagement:

- Top-up rate: **91.20%**
- Transaction-active rate: **88.91%**

Engagement declined through 2021–2023, with the **2023 cohort** recording:

- Top-up rate: **78.60%**
- Transaction-active rate: **75.79%**

The 2024 cohort showed a temporary improvement, while the 2025 and 2026 cohorts again showed lower engagement.

The 2025 and 2026 cohorts should be interpreted cautiously because newer accounts have had less time to progress through the customer lifecycle.

**Key takeaway:**  
> The analysis suggests weaker engagement among more recent cohorts, although part of this difference may be explained by the shorter time available for newer accounts to mature.

---

# 🎯 Overall Findings

The analysis suggests that the customer journey is generally successful once customers become engaged. Most accounts complete a top-up and become transaction-active, and the majority eventually progress to repeated transactions.

However, the analysis also identifies several important opportunities:

1. **Early customer activation is important.** A meaningful percentage of accounts never reach the first top-up stage.
2. **Faster first top-up is associated with stronger transaction engagement.**
3. **Transaction frequency is strongly associated with transaction value.**
4. **Product type does not show a meaningful difference in engagement.**
5. **Verification status is strongly associated with top-up completion.**
6. **Recent cohorts show weaker engagement than the earliest cohorts, although cohort maturity must be considered.**

Overall, the strongest theme emerging from the EDA is that **early customer engagement and progression through the lifecycle are closely connected with later transaction activity and customer value**.

---

# 💡 Recommendations

Based on the findings, the following areas could be considered for further investigation or business action:

### 1. Focus on first-top-up conversion

Investigate why a proportion of newly created accounts do not complete their first top-up and identify opportunities to improve early customer activation.

### 2. Encourage earlier first top-up

Since earlier top-up timing is associated with transaction activity, onboarding or customer communication strategies could focus on helping customers complete their first top-up sooner.

### 3. Investigate verification-related drop-off

The large difference between verified and unverified accounts suggests that verification-related barriers may be an important area for further investigation.

### 4. Monitor engagement across newer cohorts

Recent cohorts should be monitored as they mature to determine whether their lower engagement levels persist or improve over time.

### 5. Focus retention efforts on transaction activity

Since transaction count has a strong relationship with total transaction value, encouraging continued transaction activity may be important for increasing customer value.

---

# ⚠️ Analysis Limitations

The findings should be interpreted within the limitations of the dataset and analysis.

- The dataset contains anonymised account-level records and does not include detailed demographic or behavioural information that could explain why customers behave differently.
- `card_activation_date` was excluded because it was identified as unreliable during data validation.
- **134 records** contained negative durations between first top-up and first transaction and were flagged as potentially inconsistent.
- Account status is highly imbalanced, with the majority of accounts classified as verified.
- Recent cohorts, particularly 2025 and 2026, have had less time to mature.
- Correlation and association findings do not establish causation.

This analysis provides an overall view of customer account engagement from account creation through top-up and transaction activity.

The results indicate that while most customers progress successfully into transaction activity, the early stages of the customer journey remain an important area for improvement. Earlier top-up behaviour is associated with stronger transaction engagement, while transaction frequency has a strong positive relationship with total transaction value.

Product type shows little difference in engagement, whereas verification status shows a much stronger association with top-up completion. Cohort analysis also indicates that engagement has generally weakened among more recent account-creation cohorts, although these newer cohorts require additional time before firm conclusions can be drawn.

Overall, the analysis highlights **early activation, continued transaction engagement, and cohort-level customer behaviour** as the key areas that can help understand and improve customer engagement.
